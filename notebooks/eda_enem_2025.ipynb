#### Análise sobre os participantes

```python
df_partcipantes = pd.read_csv("../data/PARTICIPANTES_2025.csv", sep=";", encoding="iso-8859-1")
df_partcipantes.head()
```

```python
# Quantos participantes realizaram o ENEM 2025?
qtde_participantes = len(df_partcipantes)
print(f"Quantidade total de participantes: {qtde_participantes:,}")
```

```python
# Qual a distribuição de participantes por Estado?
(df_partcipantes["SG_UF_PROVA"]
 .value_counts()
 .sort_index()
 )
```

```python
# Qual é a faixa etária dos participantes?

# Mapear os código de faixa etária
faixa_etaria = {1: "Menor de 17 anos",
                2: "17 anos",
                3: "18 anos", 
                4: "19 anos",
                5: "20 anos",
                6: "21 anos",
                7: "22 anos",
                8: "23 anos",
                9: "24 anos",
                10: "25 anos",
                11: "Entre 26 e 30 anos",
                12: "Entre 31 e 35 anos", 
                13: "Entre 36 e 40 anos",
                14: "Entre 41 e 45 anos",
                15: "Entre 46 e 50 anos",
                16: "Entre 51 e 55 anos",
                17: "Entre 56 e 60 anos",
                18: "Entre 61 e 65 anos",
                19: "Entre 66 e 70 anos",
                20: "Maior de 70 anos",
                }

df_partcipantes["TX_FAIXA_ETARIA"] = df_partcipantes["TP_FAIXA_ETARIA"].map(faixa_etaria)

(df_partcipantes[["TP_FAIXA_ETARIA", "TX_FAIXA_ETARIA"]]
 .groupby(["TP_FAIXA_ETARIA", "TX_FAIXA_ETARIA"], as_index=False)
 .value_counts()
 .sort_values(by="TP_FAIXA_ETARIA")
 )
```

```python
# Qual é a quantidade de participantes por sexo?
df_sexo = df_partcipantes[["TP_SEXO"]].groupby(by="TP_SEXO", as_index=False).value_counts()
df_sexo["percentual"] = round(df_sexo['count'] / qtde_participantes * 100, 2)
df_sexo
```

```python
# Qual é a quantidade de participantes por raça?

raca = {0: "Não declarado",
        1: "Branca",
        2: "Preta",
        3: "Parda",
        4: "Amarela",
        5: "Indígena",
        }

df_partcipantes["TX_COR_RACA"] = df_partcipantes["TP_COR_RACA"].map(raca)

df_raca = (df_partcipantes[["TP_COR_RACA", "TX_COR_RACA"]]
           .groupby(["TP_COR_RACA", "TX_COR_RACA"], as_index=False)
           .value_counts()
           .sort_values("TP_COR_RACA")
           )
df_raca["percentual"] = round(df_raca["count"] / qtde_participantes * 100, 2)
df_raca
```

```python
# Quantos treineiros realizaram o ENEM 2025?
treineiro = {0: "Não", 1: "Sim"}

df_partcipantes["TX_TREINEIRO"] = df_partcipantes["IN_TREINEIRO"].map(treineiro)
df_treineiro = (df_partcipantes[["IN_TREINEIRO", "TX_TREINEIRO"]]
                .groupby(by=["IN_TREINEIRO", "TX_TREINEIRO"], as_index=False)
                .value_counts()
                )
df_treineiro["percentual"] = round(df_treineiro["count"] / qtde_participantes * 100, 2)
df_treineiro
```
